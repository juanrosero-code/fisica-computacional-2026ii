```python
%%writefile hmaxima.f90

program hmaxima

    implicit none
      ! Defino variables y parámetros

    real :: v0, angulo_rad, angulo_grados, h_max, g
    real, parameter :: pi = 3.14159265

      ! Valor de la gravedad
    g = 9.8

      ! Datos iniciales
    print *, "Ingrese la velocidad inicial (m/s):"
    read *, v0

    print *, "Ingrese el angulo de lanzamiento (grados):"
    read *, angulo_grados

    ! Cálculo de la altura máxima

    angulo_rad = angulo_grados * pi / 180.0
    h_max = (v0**2 * sin(angulo_rad)**2) / (2.0*g)

      ! Muestra de resultados
    print *, "La altura maxima es:", h_max, "metros"

end program hmaxima
```

```python
!gfortran hmaxima.f90 -o hmaxima
```
